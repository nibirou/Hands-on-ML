# 第十二章 决策树模型

- 决策树模型采用了与树类似的结构
- 在决策树中 ，最上面的节点称为根节点，最下面没有分叉的节点称为叶节点。其中，根节点和内部节点都有些边指向其他节点，
- 这些被指向的节点就称为它的 节点。叶节点是树的最末端，没有指向其他节点的边。
- 而除根节点之外，每个内部节点和叶节点都有唯一的节点指向它，该节点称为其父节点。
- 决策树通常用于分类

## 12.1 决策树的构造
- 目标是使得最终构造的决策树的复杂度尽可能低，即判断一个样本的类别需要的期望次数尽可能少
- 考虑信息增益（information gain），在每一步（根节点、内部节点）分类时，应当选择能获得更大信息增益的特征作为分类特征。
- 通过信息熵和信息增益率（information gain rate）来判断分类特征的选择。
- 综上所述 我们可以通过在每个节点上选取信息增益率最大的分类特征来构建决策树 从而使其整体复杂度较低

## 12.2 ID3和C4.5算法
- 1979 年提出的 ID3 (iterative dichotomiser 3)算法是最早的基于信息增益进行决策树构造的算法之一，同时也是后续大多数决策树算法的基础。
- ID3 算法的流程与我们上面所说的基本相同。从根节点开始 每次选取使信息增益 l(X Y) 最大的特征进行分类 并对产生的子节点递归进行特征选取和分类（进而分裂出新节点），直到所有节点上的数据都属同一类别为止。
- 由于 ID3 算法倾向于无限精细划分 它存在两个较为严重的问题。第一个问题是 如果用于划分的特征本 就较为复杂 就会出现在 12.1 节中提到的复杂度转移现象。 另一方面，继续分类的复杂度只 转移到了分支选择上 另一方面 这样的精细划分极有可能只对当前的训练集适用 ，容易出现过拟合现象。
